# BINFA TP6 : Apprentissage Hebbien et réseau de Hopfield

Les réseaux de Hopfield implémentent le concept de mémoire associative. Les motifs sont stockés comme des attracteurs du réseau permettant de retrouver un motif même à partir d'un fragment de celui-ci.

Un réseau de Hopfield est composé de $N$ neurones, chaque neurone $i=1..N$ est connecté à chaque autre neurone $j=1..N$ via un poids symmétrique $W_{ij} = W_{ji}$. Les neurones n'ont pas de connections avec eux-même donc $W_{ii} = 0$. Finalement chaque neurone peut être dans deux état $\{-1, +1\}$, $-1$ signifiant inactif et $+1$ actif.

Un réseau à $N$ neurones peut uniquement stocker des motifs de taille $N$. Un motif $\xi$ est simplement un vecteur de taille $N$. Ces motifs vont être stockés dans les poids du réseau.

## Etape 1 : Apprentissage

Pour stocker $P$ motifs $\xi^{(0)}, ..., \xi^{(P-1)}$ dans un réseau de Hopfield de taille $N$ on va construire la matrice de poids $W$ via la loi d'apprentissage d'Hebb :
$$
W_{i,j} = \frac{1}{P}\sum\limits_{u=1}^{P} \xi_i^{(u)}\xi_j^{(u)}
$$

Une valeur $W_{i,j}$ positive indique une corrélation entre les valeurs de neurones $i$ et $j$, une valeur neutre indique une indépendance et une valeur négative indique une anticorrélation.

Dans la formule on voit que:
* Si $\xi^{(u)}_i = \xi^{(u)}_j$ alors $\xi^{(u)}_i\xi^{(u)}_j > 0$
* Si $\xi^{(u)}_i \neq \xi^{(u)}_j$ alors $\xi^{(u)}_i\xi^{(u)}_j < 0$

Finalement, après avoir calculer $W_{ij}$ il faut s'assurer que $W_{ii} = 0$.
